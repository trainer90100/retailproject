```python
from pyspark.sql import *
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
data = [("Alice", 10), ("Bob", 12), ("Charlie", 15)]
schema = ["name","id"]
df = spark.createDataFrame(data, schema)
display(df)
```

```python
data = [("Alice", 10), ("Bob", 12), ("Charlie", 15)]
schema = StructType([StructField("name", StringType(), False), 
                     StructField("age", IntegerType(), False)])
df = spark.createDataFrame(data, schema)
display(df)

```

```python
data = [("Alice", 10), ("Bob", 12), ("Charlie", 15)]
schema = ["name","id"]
df = spark.createDataFrame(data, schema)
display(df)
```

```python
data = [
    [1, "Amit Sharma", 55000],
    [2, "Priya Verma", 62000],
    [3, "Rahul Mehta", 58000],
    [4, "Sneha Iyer", 60000],
    [5, "Vikram Singh", 75000],
    [6, "Neha Reddy", 67000],
    [7, "Rohan Das", 59000],
    [8, "Kavita Joshi", 53000],
    [9, "Arjun Nair", 71000],
    [10, "Pooja Saxena", 65000]
]
schema = ["id", "name", "salary"]
df1 = spark.createDataFrame(data, schema)
display(df1)
```

```python
# broadcast join
df_join = df1.join(broadcast(df),df1.id==df.id, "inner")
display(df_join)
```

```python
BV = broadcast(df)
df_join = df1.join(BV,df1.id==BV.id, "inner")
df_join.show()
```
